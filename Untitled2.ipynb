```python
column_names = ['target', 'ids', 'date', 'flag', 'user', 'text']
twitter_data = pd.read_csv('/content/training.1600000.processed.noemoticon.csv',names=column_names, encoding='ISO-8859-1', engine='python', on_bad_lines='warn')
```

```python
twitter_data.replace({'target':{4:1}}, inplace=True)
```

```python
twitter_data['target'].value_counts()
```

0 -----> negative Tweet

1 -----> positive Tweet


Stemming


```python
port_stem = PorterStemmer()
```

```python
def stemming(content):
  stemmed_content = re.sub('[^a-zA-Z]',' ',content)
  stemmed_content = stemmed_content.lower()
  stemmed_content = stemmed_content.split()
  stemmed_content = [port_stem.stem(word) for word in stemmed_content if not word in stopwords.words('english')]
  stemmed_content = ' '.join(stemmed_content)
  return stemmed_content

```

```python
twitter_data['stemmed_content'] = twitter_data['text'].apply(stemming)
```

```python
twitter_data.head()
```

```python
print(twitter_data['stemmed_content'])
```

```python
print(twitter_data['target'])
```

```python
X = twitter_data['stemmed_content'].values
Y = twitter_data['target'].values
```

```python
print(X)
```

```python
print(Y)
```

SPLITTING DATA INTO TEST DATA


```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print(X_train)
```

```python
print(X_test)
```

```python
vectorizer = TfidfVectorizer()

X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)
```

```python
print(X_train)
```

```python
print(X_test)
```

training machine learning model

logistic regression


```python
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train, Y_train)
```

model evaluation

```python
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy score of the training data : ', training_data_accuracy)
```

```python
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy score of the test data : ', test_data_accuracy)
```

Model accuaracy = 77.8%

using saved model for future predicting

```python
loaded_model = pickle.load(open('/content/trained_model.sav', 'rb'))
```

```python
input_tweet = input('Enter your tweet: ')

# Preprocess the input tweet
stemmed_input = stemming(input_tweet)

# Vectorize the stemmed input tweet
vectorized_input = vectorizer.transform([stemmed_input])

# Make prediction
prediction = loaded_model.predict(vectorized_input)

if (prediction[0] == 0):
  print('Negative 😞')
else:
  print('Positive 😊')
```

### Testing the Model with a Batch of Tweets from the DataFrame

```python
# Select a small batch of tweets from the original dataframe for testing
# Let's pick 5 random tweets, or specify indices if preferred
sample_tweets_df = twitter_data.sample(5, random_state=42)

print("Original Sample Tweets and their true targets (if available):")
for index, row in sample_tweets_df.iterrows():
    print(f"ID: {row['ids']}, Text: '{row['text']}', True Target: {row['target']}")

# Preprocess the sample tweets
stemmed_sample_tweets = sample_tweets_df['text'].apply(stemming)

# Vectorize the stemmed sample tweets
vectorized_sample_tweets = vectorizer.transform(stemmed_sample_tweets)

# Make predictions
sample_predictions = loaded_model.predict(vectorized_sample_tweets)

print("\nPredictions for the Sample Tweets:")
for i, (original_text, prediction) in enumerate(zip(sample_tweets_df['text'], sample_predictions)):
    sentiment = 'Positive' if prediction == 1 else 'Negative'
    print(f"Tweet {i+1}: '{original_text}' -> {sentiment}")
```
